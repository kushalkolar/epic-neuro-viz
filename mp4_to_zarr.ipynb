```python
cap = cv2.VideoCapture(path)
t = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
m = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
n = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

shape = (t, m, n)
shape
```
